```python
from itertools import count

import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Ex_Files_Introducing_Jupyter/Ex_Files_Introducing_Jupyter/Exercise Files/inputs/Environmental_Data_Deep_Moor_2015.csv')
```

# Number of measurements

```python
feb_records = df[df['date'].between('2015_01_01', '2015_01_12')]
```

```python
feb_records
```

```python
dates_in_feb = feb_records['date'].unique()
```

```python
counts = df['date'].value_counts()
records_by_day = [counts[x] for x in dates_in_feb]
```

```python
plt.plot(dates_in_feb, records_by_day)
```
